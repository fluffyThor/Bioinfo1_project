## Step 2. Gene-level master table 만들기

이 단계의 목표:
- read count table을 다시 불러온다.
- sample 이름을 정리한다.
- 각 sample의 library size를 확인한다.
- CPM으로 normalization한다.
- RNA 변화, RPF 변화, translation efficiency 변화를 계산한다.

주요 계산:
- log2FC_RNA = log2(RNA-siLin28a / RNA-siLuc)
- log2FC_RPF = log2(RPF-siLin28a / RPF-siLuc)
- delta_TE = log2FC_RPF - log2FC_RNA
- CLIP_enrichment = log2(CLIP-35L33G / RNA-control)

````python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 네가 알려준 프로젝트 폴더 경로
DATA_DIR = Path("/home/ssy010622/bioinfo1_project/binfo1-work")

print("DATA_DIR:", DATA_DIR)
print("exists:", DATA_DIR.exists())
````

````python
count_file = DATA_DIR / "read-counts_multimap.txt"

print("count_file:", count_file)
print("exists:", count_file.exists())
````

````python
count_file = DATA_DIR / "read-counts_multimap.txt"

summary_candidates = sorted(DATA_DIR.glob("*multimap*.summary"))
````

````python
counts_raw = pd.read_csv(count_file, sep="\t", comment="#")

print("raw shape:", counts_raw.shape)
print("raw columns:")
print(counts_raw.columns.tolist())

counts = counts_raw.copy()

# sample column 이름에서 .bam 제거
rename_map = {}
for c in counts.columns:
    clean = Path(c).name
    clean = clean.replace(".bam", "")
    rename_map[c] = clean

counts = counts.rename(columns=rename_map)

print("\ncleaned columns:")
print(counts.columns.tolist())

counts.head()
````

````python
main_samples = [
    "CLIP-35L33G",
    "RNA-control",
    "RNA-siLin28a",
    "RNA-siLuc",
    "RPF-siLin28a",
    "RPF-siLuc"
]

# 혹시 빠진 column이 있는지 확인
missing = [s for s in main_samples if s not in counts.columns]

if len(missing) == 0:
    print("모든 main sample column이 존재합니다.")
else:
    print("빠진 sample column:", missing)
````

````python
# count column을 숫자형으로 변환
for s in main_samples:
    counts[s] = pd.to_numeric(counts[s], errors="coerce").fillna(0)

# sample별 assigned count 총합 계산
library_size = counts[main_samples].sum()

library_table = pd.DataFrame({
    "sample": library_size.index,
    "assigned_count": library_size.values,
    "assigned_count_million": library_size.values / 1_000_000
})

library_table
````

````python
if len(summary_candidates) > 0:
    summary_file = summary_candidates[0]
    print("사용할 summary file:", summary_file.name)

    summary = pd.read_csv(summary_file, sep="\t", index_col=0)

    print("summary shape:", summary.shape)
    display(summary.head())

    if "Assigned" in summary.columns:
        summary_assigned = summary["Assigned"].copy()
        summary_assigned.index = summary_assigned.index.str.replace(".bam", "", regex=False)

        compare_library = pd.concat(
            [
                library_size.rename("sum_from_count_table"),
                summary_assigned.rename("assigned_from_summary")
            ],
            axis=1
        )

        compare_library["difference"] = (
            compare_library["sum_from_count_table"] 
            - compare_library["assigned_from_summary"]
        )

        display(compare_library)
else:
    print("summary 파일을 찾지 못했습니다.")
````

````python
# QC 용 

library_table_plot = library_table.set_index("sample")

plt.figure(figsize=(7, 4))
library_table_plot["assigned_count_million"].plot(kind="bar")

plt.ylabel("Assigned counts (million)")
plt.xlabel("Sample")
plt.title("Library size based on assigned gene counts")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
````

````python
meta_cols = ["Geneid", "Chr", "Start", "End", "Strand", "Length"]

master = counts[meta_cols + main_samples].copy()

# Geneid에서 version 제거한 column도 만들어두기
# 예: ENSMUSG00000000001.4 -> ENSMUSG00000000001
master["gene_id_no_version"] = master["Geneid"].str.replace(r"\.\d+$", "", regex=True)

master.head()
````

````python
for s in main_samples:
    master[f"{s}_CPM"] = master[s] / library_size[s] * 1_000_000

cpm_cols = [f"{s}_CPM" for s in main_samples]

master[["Geneid", "Length"] + cpm_cols].head()
````

````python
# log 계산에서 0으로 나누는 것을 막기 위한 작은 값
pseudo_cpm = 0.5

master["log2FC_RNA"] = np.log2(
    (master["RNA-siLin28a_CPM"] + pseudo_cpm) /
    (master["RNA-siLuc_CPM"] + pseudo_cpm)
)

master["log2FC_RPF"] = np.log2(
    (master["RPF-siLin28a_CPM"] + pseudo_cpm) /
    (master["RPF-siLuc_CPM"] + pseudo_cpm)
)

# Translation efficiency 변화
# RPF 변화에서 RNA abundance 변화를 뺀 값
master["delta_TE"] = master["log2FC_RPF"] - master["log2FC_RNA"]

# LIN28A CLIP enrichment
# CLIP signal을 RNA-control abundance로 보정
master["CLIP_enrichment"] = np.log2(
    (master["CLIP-35L33G_CPM"] + pseudo_cpm) /
    (master["RNA-control_CPM"] + pseudo_cpm)
)

master[
    [
        "Geneid",
        "Length",
        "RNA-siLuc_CPM",
        "RNA-siLin28a_CPM",
        "RPF-siLuc_CPM",
        "RPF-siLin28a_CPM",
        "CLIP-35L33G_CPM",
        "RNA-control_CPM",
        "log2FC_RNA",
        "log2FC_RPF",
        "delta_TE",
        "CLIP_enrichment"
    ]
].head()
````

````python
analysis_df = master[
    (master["RNA-siLuc_CPM"] >= 1) &
    (master["RNA-siLin28a_CPM"] >= 1) &
    ((master["RPF-siLuc"] + master["RPF-siLin28a"]) >= 20)
].copy()

print("전체 gene 수:", len(master))
print("필터 통과 gene 수:", len(analysis_df))

analysis_df[
    [
        "Geneid",
        "RNA-siLuc_CPM",
        "RNA-siLin28a_CPM",
        "RPF-siLuc_CPM",
        "RPF-siLin28a_CPM",
        "log2FC_RNA",
        "log2FC_RPF",
        "delta_TE",
        "CLIP_enrichment"
    ]
].head()
````

### Filtering decision

본 분석의 목적은 LIN28A가 mRNA translation, 특히 ER-associated translation에 미치는 영향을 보는 것이다. 따라서 rRNA/tRNA 및 mitochondrial genome-encoded features는 main analysis에서 제외하였다. 그러나 nuclear-encoded mitochondrial proteins는 일반적인 protein-coding mRNA이므로 제거하지 않았다.

논문에서도 mRNA/RefSeq transcript 중심으로 CLIP enrichment와 ribosome density change를 분석했으며, rRNA/tRNA reads는 subsequent analysis에서 제외하였다. 또한 histone mRNA는 poly(A) RNA-seq normalization 문제로 CLIP enrichment가 과대평가될 수 있다고 언급되어 있어, histone 제외 여부를 sensitivity analysis로 확인한다.

````python
# 전체 protein-coding set: chrM 포함
analysis_pc_all = master_annot[
    (master_annot["gene_type"] == "protein_coding") &
    (master_annot["RNA-siLuc_CPM"] >= 1) &
    (master_annot["RNA-siLin28a_CPM"] >= 1) &
    ((master_annot["RPF-siLuc"] + master_annot["RPF-siLin28a"]) >= 20)
].copy()

# chrM 제외
analysis_pc_no_chrM = analysis_pc_all[
    ~analysis_pc_all["Chr"].astype(str).str.contains("chrM", na=False)
].copy()

# histone gene 표시
# mouse GENCODE symbol에서 H1f, H2a/H2b/H3/H4 계열을 넓게 잡음
histone_pattern = r"^(Hist|H1f|H2a|H2b|H3|H4|H2ac|H2bc|H3c|H4c)"

analysis_pc_no_chrM["is_histone"] = analysis_pc_no_chrM["gene_name"].astype(str).str.contains(
    histone_pattern,
    regex=True,
    case=False,
    na=False
)

# chrM + histone 제외
analysis_pc_no_chrM_no_histone = analysis_pc_no_chrM[
    ~analysis_pc_no_chrM["is_histone"]
].copy()

print("protein-coding, chrM 포함:", len(analysis_pc_all))
print("protein-coding, chrM 제외:", len(analysis_pc_no_chrM))
print("protein-coding, chrM + histone 제외:", len(analysis_pc_no_chrM_no_histone))



def make_corr_table(df, label):
    rows = []
    pairs = [
        ("CLIP_enrichment", "log2FC_RNA"),
        ("CLIP_enrichment", "log2FC_RPF"),
        ("CLIP_enrichment", "delta_TE"),
    ]

    for x_col, y_col in pairs:
        temp = df[[x_col, y_col]].replace([np.inf, -np.inf], np.nan).dropna()

        rows.append({
            "set": label,
            "x": x_col,
            "y": y_col,
            "pearson": temp[x_col].corr(temp[y_col], method="pearson"),
            "spearman": temp[x_col].corr(temp[y_col], method="spearman"),
            "n_genes": len(temp)
        })

    return pd.DataFrame(rows)

corr_compare = pd.concat(
    [
        make_corr_table(analysis_pc_all, "protein_coding_all_chr"),
        make_corr_table(analysis_pc_no_chrM, "protein_coding_no_chrM"),
        make_corr_table(analysis_pc_no_chrM_no_histone, "protein_coding_no_chrM_no_histone"),
    ],
    ignore_index=True
)

corr_compare
````

## Step 7. LIN28A binding strength별 binning 분석

````python
def make_corr_table(df, label):
    rows = []
    pairs = [
        ("CLIP_enrichment", "log2FC_RNA"),
        ("CLIP_enrichment", "log2FC_RPF"),
        ("CLIP_enrichment", "delta_TE"),
    ]

    for x_col, y_col in pairs:
        temp = df[[x_col, y_col]].replace([np.inf, -np.inf], np.nan).dropna()

        rows.append({
            "set": label,
            "x": x_col,
            "y": y_col,
            "pearson": temp[x_col].corr(temp[y_col], method="pearson"),
            "spearman": temp[x_col].corr(temp[y_col], method="spearman"),
            "n_genes": len(temp)
        })

    return pd.DataFrame(rows)

corr_compare2 = pd.concat(
    [
        make_corr_table(analysis_main, "main_no_chrM"),
        make_corr_table(analysis_no_histone, "no_chrM_no_histone"),
        make_corr_table(analysis_paperlike, "paperlike"),
        make_corr_table(analysis_paperlike_no_histone, "paperlike_no_histone"),
    ],
    ignore_index=True
)

corr_compare2
````

### Interpretation

Filtering sensitivity analysis showed that the positive association between LIN28A CLIP enrichment and delta_TE was robust to removal of chrM genes and histone genes. Therefore, the observed relationship is unlikely to be driven solely by mitochondrial genome-encoded features or histone-related normalization artifacts.

Next, genes were divided into four groups according to LIN28A CLIP enrichment: bottom 50%, 20–50%, 5–20%, and top 5%. This follows the strategy used in the original paper. If the top 5% LIN28A binders show higher delta_TE than weak binders, it supports the idea that LIN28A-bound mRNAs are translationally derepressed after Lin28a knockdown.

### Result 1. LIN28A binding strength predicts translational derepression

Genes were divided into four groups based on LIN28A CLIP enrichment. The cumulative distribution of delta_TE shifted progressively to the right as LIN28A binding strength increased. The top 5% LIN28A-bound genes showed the highest median delta_TE, whereas weak binders showed the lowest median delta_TE.

This reproduces the central pattern of the original study: LIN28A-bound mRNAs tend to show increased ribosome occupancy / translation efficiency after Lin28a knockdown. Therefore, LIN28A appears to act mainly as a translational repressor for its bound mRNAs rather than as a simple stabilizer of mRNA abundance.

````python
def keyword_flag(gene_name):
    name = str(gene_name)
    
    patterns = {
        "Tmem_like": r"^(Tmem)",
        "Transporter_like": r"^(Slc|Abc)",
        "Collagen_ECM_like": r"^(Col|Lam|Fn|Nid|Mmp|Itg)",
        "Lysosome_membrane_like": r"^(Lamp)",
        "ER_Golgi_like": r"^(Sec|Tmx|Pdia|Hsp90b1|Hspa5|Canx|Calr|Erlec|Alg|Bcap|Surf)",
        "Receptor_surface_like": r"^(Tlr|Cd|Cdh|Epcam|Gpr)"
    }
    
    hits = []
    for label, pat in patterns.items():
        if pd.Series([name]).str.contains(pat, regex=True, case=False).iloc[0]:
            hits.append(label)
    
    if len(hits) == 0:
        return "no_keyword_hit"
    else:
        return ";".join(hits)

functional_targets["keyword_category"] = functional_targets["gene_name"].apply(keyword_flag)
top5["keyword_category"] = top5["gene_name"].apply(keyword_flag)
df["keyword_category"] = df["gene_name"].apply(keyword_flag)

functional_targets["keyword_category"].value_counts()
````

## Step 3. Main hypothesis group setup

논문 재현 분석에서는 LIN28A binding strength가 높을수록 delta_TE가 증가한다는 것을 확인했다.

이제부터의 질문은 조금 다르다.

Main question:
LIN28A가 강하게 결합한 mRNA들 중에서, 실제 translation derepression이 큰 gene과 작은 gene은 무엇이 다른가?

Hypothesis:
LIN28A binding 자체만으로는 functional repression 강도를 설명하기 어렵고,
strong LIN28A-bound mRNA 중에서도 ER/membrane/secretory/lysosomal context를 가진 gene들이 더 큰 delta_TE를 보일 것이다.

````python
from pathlib import Path
import pandas as pd
import numpy as np

DATA_DIR = Path("/home/ssy010622/bioinfo1_project/binfo1-work")
REGION_DIR = DATA_DIR / "region_counts"

cds_out = REGION_DIR / "CLIP_CDS_counts.txt"
utr_out = REGION_DIR / "CLIP_UTR_counts.txt"

print("CDS count file exists:", cds_out.exists())
print("UTR count file exists:", utr_out.exists())
````

````python
def read_featurecounts_count(count_file, region_name):
    """
    Read featureCounts output and return gene-level count + feature length.
    """
    raw = pd.read_csv(count_file, sep="\t", comment="#")
    
    # featureCounts output columns:
    # Geneid, Chr, Start, End, Strand, Length, sample_count_column
    count_col = raw.columns[-1]
    
    out = raw[["Geneid", "Length", count_col]].copy()
    out = out.rename(columns={
        "Length": f"{region_name}_length",
        count_col: f"{region_name}_CLIP_count"
    })
    
    out["gene_id_no_version"] = out["Geneid"].astype(str).str.replace(
        r"\.\d+$", "", regex=True
    )
    
    return out

cds_region = read_featurecounts_count(cds_out, "CDS")
utr_region = read_featurecounts_count(utr_out, "UTR")

print("CDS region table:")
display(cds_region.head())

print("UTR region table:")
display(utr_region.head())
````

````python
# df가 없으면 저장된 binned table에서 불러오기
try:
    df
    print("기존 df를 사용합니다.")
except NameError:
    binned_file = DATA_DIR / "02_paperlike_no_histone_binned.tsv"
    df = pd.read_csv(binned_file, sep="\t")
    print("df를 파일에서 불러왔습니다:", binned_file)

# gene_id_no_version이 없으면 만들기
if "gene_id_no_version" not in df.columns:
    df["gene_id_no_version"] = df["Geneid"].astype(str).str.replace(
        r"\.\d+$", "", regex=True
    )

region_df = df.merge(
    cds_region[["gene_id_no_version", "CDS_length", "CDS_CLIP_count"]],
    on="gene_id_no_version",
    how="left"
).merge(
    utr_region[["gene_id_no_version", "UTR_length", "UTR_CLIP_count"]],
    on="gene_id_no_version",
    how="left"
)

# missing count는 0으로 처리
for col in ["CDS_CLIP_count", "UTR_CLIP_count", "CDS_length", "UTR_length"]:
    region_df[col] = region_df[col].fillna(0)

print("original df shape:", df.shape)
print("region_df shape:", region_df.shape)

region_df[
    [
        "Geneid",
        "gene_name",
        "CLIP_enrichment",
        "delta_TE",
        "CDS_length",
        "UTR_length",
        "CDS_CLIP_count",
        "UTR_CLIP_count"
    ]
].head()
````

````python
pseudo = 0.5

region_df["CDS_UTR_CLIP_total"] = (
    region_df["CDS_CLIP_count"] + region_df["UTR_CLIP_count"]
)

region_df["CDS_CLIP_fraction"] = np.where(
    region_df["CDS_UTR_CLIP_total"] > 0,
    region_df["CDS_CLIP_count"] / region_df["CDS_UTR_CLIP_total"],
    np.nan
)

region_df["UTR_CLIP_fraction"] = np.where(
    region_df["CDS_UTR_CLIP_total"] > 0,
    region_df["UTR_CLIP_count"] / region_df["CDS_UTR_CLIP_total"],
    np.nan
)

# 단순 count bias: CDS count가 UTR count보다 얼마나 큰가
region_df["CDS_vs_UTR_CLIP_bias"] = np.log2(
    (region_df["CDS_CLIP_count"] + pseudo) /
    (region_df["UTR_CLIP_count"] + pseudo)
)

# 길이 보정 density: counts per kb
region_df["CDS_length_kb"] = region_df["CDS_length"] / 1000
region_df["UTR_length_kb"] = region_df["UTR_length"] / 1000

region_df["CDS_CLIP_density_per_kb"] = np.where(
    region_df["CDS_length_kb"] > 0,
    region_df["CDS_CLIP_count"] / region_df["CDS_length_kb"],
    np.nan
)

region_df["UTR_CLIP_density_per_kb"] = np.where(
    region_df["UTR_length_kb"] > 0,
    region_df["UTR_CLIP_count"] / region_df["UTR_length_kb"],
    np.nan
)

region_df["CDS_vs_UTR_density_bias"] = np.log2(
    (region_df["CDS_CLIP_density_per_kb"] + pseudo) /
    (region_df["UTR_CLIP_density_per_kb"] + pseudo)
)

region_df[
    [
        "gene_name",
        "CLIP_enrichment",
        "delta_TE",
        "CDS_CLIP_count",
        "UTR_CLIP_count",
        "CDS_CLIP_fraction",
        "CDS_vs_UTR_CLIP_bias",
        "CDS_CLIP_density_per_kb",
        "UTR_CLIP_density_per_kb",
        "CDS_vs_UTR_density_bias"
    ]
].head()
````

````python
# top5 strong binder가 이미 effect_group을 가지고 있는지 확인
top5_region = region_df[region_df["LIN28A_binding_group"] == "top 5%"].copy()

# effect_group이 없다면 다시 만들기
if "effect_group" not in top5_region.columns:
    q25 = top5_region["delta_TE"].quantile(0.25)
    q75 = top5_region["delta_TE"].quantile(0.75)
    
    top5_region["effect_group"] = "middle"
    top5_region.loc[top5_region["delta_TE"] <= q25, "effect_group"] = "low_delta_TE"
    top5_region.loc[top5_region["delta_TE"] >= q75, "effect_group"] = "high_delta_TE"

top5_region["effect_group"] = pd.Categorical(
    top5_region["effect_group"],
    categories=["low_delta_TE", "middle", "high_delta_TE"],
    ordered=True
)

region_feature_summary = top5_region.groupby("effect_group").agg(
    n_genes=("Geneid", "count"),
    median_delta_TE=("delta_TE", "median"),
    median_CLIP_enrichment=("CLIP_enrichment", "median"),
    median_CDS_CLIP_count=("CDS_CLIP_count", "median"),
    median_UTR_CLIP_count=("UTR_CLIP_count", "median"),
    median_CDS_CLIP_fraction=("CDS_CLIP_fraction", "median"),
    median_CDS_vs_UTR_CLIP_bias=("CDS_vs_UTR_CLIP_bias", "median"),
    median_CDS_CLIP_density_per_kb=("CDS_CLIP_density_per_kb", "median"),
    median_UTR_CLIP_density_per_kb=("UTR_CLIP_density_per_kb", "median"),
    median_CDS_vs_UTR_density_bias=("CDS_vs_UTR_density_bias", "median")
).reset_index()

region_feature_summary
````

````python
from scipy.stats import mannwhitneyu
import pandas as pd
import numpy as np

# high vs low만 비교
top5_extreme_region = top5_region[
    top5_region["effect_group"].isin(["low_delta_TE", "high_delta_TE"])
].copy()

region_features = [
    "CDS_CLIP_count",
    "UTR_CLIP_count",
    "CDS_CLIP_fraction",
    "CDS_vs_UTR_CLIP_bias",
    "CDS_CLIP_density_per_kb",
    "UTR_CLIP_density_per_kb",
    "CDS_vs_UTR_density_bias"
]

region_test_rows = []

for col in region_features:
    low_values = top5_extreme_region.loc[
        top5_extreme_region["effect_group"] == "low_delta_TE", col
    ].replace([np.inf, -np.inf], np.nan).dropna()
    
    high_values = top5_extreme_region.loc[
        top5_extreme_region["effect_group"] == "high_delta_TE", col
    ].replace([np.inf, -np.inf], np.nan).dropna()
    
    test = mannwhitneyu(high_values, low_values, alternative="two-sided")
    
    region_test_rows.append({
        "feature": col,
        "n_low": len(low_values),
        "n_high": len(high_values),
        "median_low": low_values.median(),
        "median_high": high_values.median(),
        "median_difference_high_minus_low": high_values.median() - low_values.median(),
        "MWU_pvalue": test.pvalue
    })

region_feature_tests = pd.DataFrame(region_test_rows)
region_feature_tests
````

### Refined binding-position hypothesis

The broad CDS-vs-UTR analysis did not show a strong association between CDS-biased LIN28A binding and delta_TE.

Therefore, I refine the hypothesis:

LIN28A-mediated translational repression may depend on local binding position relative to translation landmarks, such as the start codon or stop codon, rather than broad CDS/UTR binding bias.

Next, I will create custom annotation files for ±100 nt windows around start codons and stop codons, then count LIN28A CLIP reads in these regions.

````python
from pathlib import Path
import pandas as pd
import numpy as np

DATA_DIR = Path("/home/ssy010622/bioinfo1_project/binfo1-work")
GTF = DATA_DIR / "gencode.gtf"

WINDOW_DIR = DATA_DIR / "window_annotations"
WINDOW_DIR.mkdir(exist_ok=True)

print("GTF exists:", GTF.exists())
print("WINDOW_DIR:", WINDOW_DIR)
````

````python
# GTF 불러오기
gtf_cols = [
    "seqname",
    "source",
    "feature",
    "start",
    "end",
    "score",
    "strand",
    "frame",
    "attribute"
]

gtf = pd.read_csv(
    GTF,
    sep="\t",
    comment="#",
    names=gtf_cols,
    low_memory=False
)

# gene_id 추출
gtf["gene_id"] = gtf["attribute"].str.extract(r'gene_id "([^"]+)"')
gtf["gene_id_no_version"] = gtf["gene_id"].str.replace(r"\.\d+$", "", regex=True)

print(gtf["feature"].value_counts().loc[["start_codon", "stop_codon"]])
````

````python
def make_codon_window_saf(gtf, feature_name, window=100):
    """
    Make SAF annotation for ±window nt around start_codon or stop_codon.
    SAF columns required by featureCounts:
    GeneID, Chr, Start, End, Strand
    """
    codon = gtf[gtf["feature"] == feature_name].copy()
    
    # 혹시 gene_id가 없는 행 제거
    codon = codon.dropna(subset=["gene_id"])
    
    # codon feature가 3nt이므로 그 주변 window 생성
    codon["window_start"] = (codon["start"] - window).clip(lower=1)
    codon["window_end"] = codon["end"] + window
    
    saf = codon[["gene_id", "seqname", "window_start", "window_end", "strand"]].copy()
    saf = saf.rename(columns={
        "gene_id": "GeneID",
        "seqname": "Chr",
        "window_start": "Start",
        "window_end": "End",
        "strand": "Strand"
    })
    
    # featureCounts는 정수 좌표 필요
    saf["Start"] = saf["Start"].astype(int)
    saf["End"] = saf["End"].astype(int)
    
    # 같은 gene에 isoform별 start/stop window가 여러 개 있을 수 있음
    # 일단 EDA에서는 유지한다.
    return saf

start100_saf = make_codon_window_saf(gtf, "start_codon", window=100)
stop100_saf = make_codon_window_saf(gtf, "stop_codon", window=100)

print("start100_saf shape:", start100_saf.shape)
print("stop100_saf shape:", stop100_saf.shape)

display(start100_saf.head())
display(stop100_saf.head())
````

## Step 9. Start/stop-proximal LIN28A binding features

The broad CDS-vs-UTR analysis did not clearly explain high vs low delta_TE among strong LIN28A binders.

Next, I ask whether local LIN28A binding near translation landmarks is associated with functional derepression.

Features:
- start100_CLIP_count: LIN28A CLIP reads within ±100 nt of start codons
- stop100_CLIP_count: LIN28A CLIP reads within ±100 nt of stop codons
- start100_CLIP_fraction
- stop100_CLIP_fraction
- start_vs_stop_CLIP_bias
- start100_CLIP_density_per_kb
- stop100_CLIP_density_per_kb

````python
from pathlib import Path
import pandas as pd
import numpy as np

DATA_DIR = Path("/home/ssy010622/bioinfo1_project/binfo1-work")
WINDOW_COUNT_DIR = DATA_DIR / "window_counts"

start100_out = WINDOW_COUNT_DIR / "CLIP_start_codon_pm100_counts.txt"
stop100_out = WINDOW_COUNT_DIR / "CLIP_stop_codon_pm100_counts.txt"

print("start100 count exists:", start100_out.exists())
print("stop100 count exists:", stop100_out.exists())
````

````python
def read_featurecounts_window_count(count_file, prefix):
    raw = pd.read_csv(count_file, sep="\t", comment="#")
    
    # featureCounts output usually:
    # Geneid, Chr, Start, End, Strand, Length, sample_count_column
    gene_col = raw.columns[0]
    count_col = raw.columns[-1]
    
    out = raw[[gene_col, "Length", count_col]].copy()
    out = out.rename(columns={
        gene_col: "Geneid",
        "Length": f"{prefix}_length",
        count_col: f"{prefix}_CLIP_count"
    })
    
    out["gene_id_no_version"] = out["Geneid"].astype(str).str.replace(
        r"\.\d+$", "", regex=True
    )
    
    return out

start100_counts = read_featurecounts_window_count(start100_out, "start100")
stop100_counts = read_featurecounts_window_count(stop100_out, "stop100")

print("start100_counts:")
display(start100_counts.head())

print("stop100_counts:")
display(stop100_counts.head())
````

````python
# region_df가 없으면 저장된 파일에서 불러오기
try:
    region_df
    print("기존 region_df를 사용합니다.")
except NameError:
    region_file = DATA_DIR / "06_gene_table_with_CDS_UTR_CLIP_features.tsv"
    region_df = pd.read_csv(region_file, sep="\t")
    print("region_df를 파일에서 불러왔습니다:", region_file)

if "gene_id_no_version" not in region_df.columns:
    region_df["gene_id_no_version"] = region_df["Geneid"].astype(str).str.replace(
        r"\.\d+$", "", regex=True
    )

position_df = region_df.merge(
    start100_counts[["gene_id_no_version", "start100_length", "start100_CLIP_count"]],
    on="gene_id_no_version",
    how="left"
).merge(
    stop100_counts[["gene_id_no_version", "stop100_length", "stop100_CLIP_count"]],
    on="gene_id_no_version",
    how="left"
)

for col in ["start100_length", "stop100_length", "start100_CLIP_count", "stop100_CLIP_count"]:
    position_df[col] = position_df[col].fillna(0)

print("region_df shape:", region_df.shape)
print("position_df shape:", position_df.shape)

position_df[
    [
        "gene_name",
        "CLIP_enrichment",
        "delta_TE",
        "start100_length",
        "stop100_length",
        "start100_CLIP_count",
        "stop100_CLIP_count"
    ]
].head()
````

````python
pseudo = 0.5

position_df["start_stop_CLIP_total"] = (
    position_df["start100_CLIP_count"] + position_df["stop100_CLIP_count"]
)

position_df["start100_CLIP_fraction"] = np.where(
    position_df["start_stop_CLIP_total"] > 0,
    position_df["start100_CLIP_count"] / position_df["start_stop_CLIP_total"],
    np.nan
)

position_df["stop100_CLIP_fraction"] = np.where(
    position_df["start_stop_CLIP_total"] > 0,
    position_df["stop100_CLIP_count"] / position_df["start_stop_CLIP_total"],
    np.nan
)

position_df["start_vs_stop_CLIP_bias"] = np.log2(
    (position_df["start100_CLIP_count"] + pseudo) /
    (position_df["stop100_CLIP_count"] + pseudo)
)

position_df["start100_length_kb"] = position_df["start100_length"] / 1000
position_df["stop100_length_kb"] = position_df["stop100_length"] / 1000

position_df["start100_CLIP_density_per_kb"] = np.where(
    position_df["start100_length_kb"] > 0,
    position_df["start100_CLIP_count"] / position_df["start100_length_kb"],
    np.nan
)

position_df["stop100_CLIP_density_per_kb"] = np.where(
    position_df["stop100_length_kb"] > 0,
    position_df["stop100_CLIP_count"] / position_df["stop100_length_kb"],
    np.nan
)

position_df["start_vs_stop_density_bias"] = np.log2(
    (position_df["start100_CLIP_density_per_kb"] + pseudo) /
    (position_df["stop100_CLIP_density_per_kb"] + pseudo)
)

position_df[
    [
        "gene_name",
        "delta_TE",
        "start100_CLIP_count",
        "stop100_CLIP_count",
        "start100_CLIP_fraction",
        "start_vs_stop_CLIP_bias",
        "start100_CLIP_density_per_kb",
        "stop100_CLIP_density_per_kb",
        "start_vs_stop_density_bias"
    ]
].head()
````

````python
top5_position = position_df[
    position_df["LIN28A_binding_group"] == "top 5%"
].copy()

# effect_group이 없으면 다시 만들기
if "effect_group" not in top5_position.columns:
    q25 = top5_position["delta_TE"].quantile(0.25)
    q75 = top5_position["delta_TE"].quantile(0.75)
    
    top5_position["effect_group"] = "middle"
    top5_position.loc[top5_position["delta_TE"] <= q25, "effect_group"] = "low_delta_TE"
    top5_position.loc[top5_position["delta_TE"] >= q75, "effect_group"] = "high_delta_TE"

top5_position["effect_group"] = pd.Categorical(
    top5_position["effect_group"],
    categories=["low_delta_TE", "middle", "high_delta_TE"],
    ordered=True
)

startstop_feature_summary = top5_position.groupby("effect_group").agg(
    n_genes=("Geneid", "count"),
    median_delta_TE=("delta_TE", "median"),
    median_CLIP_enrichment=("CLIP_enrichment", "median"),
    median_start100_CLIP_count=("start100_CLIP_count", "median"),
    median_stop100_CLIP_count=("stop100_CLIP_count", "median"),
    median_start100_CLIP_fraction=("start100_CLIP_fraction", "median"),
    median_start_vs_stop_CLIP_bias=("start_vs_stop_CLIP_bias", "median"),
    median_start100_CLIP_density_per_kb=("start100_CLIP_density_per_kb", "median"),
    median_stop100_CLIP_density_per_kb=("stop100_CLIP_density_per_kb", "median"),
    median_start_vs_stop_density_bias=("start_vs_stop_density_bias", "median")
).reset_index()

startstop_feature_summary
````

### Result 4. Start/stop-proximal LIN28A binding analysis

Start/stop codon ±100 nt CLIP counting showed that LIN28A CLIP reads are globally more abundant near stop codons than start codons. However, high-delta_TE strong binders did not show higher start-proximal or stop-proximal CLIP signal compared with low-delta_TE strong binders.

I will statistically test whether start/stop-proximal binding features differ between high- and low-delta_TE groups.

````python
from scipy.stats import mannwhitneyu
import pandas as pd
import numpy as np

# high vs low만 비교
top5_extreme_position = top5_position[
    top5_position["effect_group"].isin(["low_delta_TE", "high_delta_TE"])
].copy()

# start/stop landmark 관련 추가 feature
top5_extreme_position["landmark100_CLIP_count"] = (
    top5_extreme_position["start100_CLIP_count"] +
    top5_extreme_position["stop100_CLIP_count"]
)

top5_extreme_position["landmark100_fraction_of_CDS_UTR"] = np.where(
    top5_extreme_position["CDS_UTR_CLIP_total"] > 0,
    top5_extreme_position["landmark100_CLIP_count"] /
    top5_extreme_position["CDS_UTR_CLIP_total"],
    np.nan
)

top5_extreme_position["start100_fraction_of_CDS_UTR"] = np.where(
    top5_extreme_position["CDS_UTR_CLIP_total"] > 0,
    top5_extreme_position["start100_CLIP_count"] /
    top5_extreme_position["CDS_UTR_CLIP_total"],
    np.nan
)

top5_extreme_position["stop100_fraction_of_CDS_UTR"] = np.where(
    top5_extreme_position["CDS_UTR_CLIP_total"] > 0,
    top5_extreme_position["stop100_CLIP_count"] /
    top5_extreme_position["CDS_UTR_CLIP_total"],
    np.nan
)

startstop_features = [
    "start100_CLIP_count",
    "stop100_CLIP_count",
    "start100_CLIP_fraction",
    "start_vs_stop_CLIP_bias",
    "start100_CLIP_density_per_kb",
    "stop100_CLIP_density_per_kb",
    "start_vs_stop_density_bias",
    "landmark100_CLIP_count",
    "landmark100_fraction_of_CDS_UTR",
    "start100_fraction_of_CDS_UTR",
    "stop100_fraction_of_CDS_UTR"
]

startstop_test_rows = []

for col in startstop_features:
    low_values = top5_extreme_position.loc[
        top5_extreme_position["effect_group"] == "low_delta_TE",
        col
    ].replace([np.inf, -np.inf], np.nan).dropna()

    high_values = top5_extreme_position.loc[
        top5_extreme_position["effect_group"] == "high_delta_TE",
        col
    ].replace([np.inf, -np.inf], np.nan).dropna()

    test = mannwhitneyu(high_values, low_values, alternative="two-sided")

    startstop_test_rows.append({
        "feature": col,
        "n_low": len(low_values),
        "n_high": len(high_values),
        "median_low": low_values.median(),
        "median_high": high_values.median(),
        "median_difference_high_minus_low": high_values.median() - low_values.median(),
        "MWU_pvalue": test.pvalue
    })

startstop_feature_tests = pd.DataFrame(startstop_test_rows)
startstop_feature_tests
````

````python
# top5_position 전체에서 delta_TE와의 상관 보기

# top5_position에도 추가 feature 만들기
top5_position["landmark100_CLIP_count"] = (
    top5_position["start100_CLIP_count"] +
    top5_position["stop100_CLIP_count"]
)

top5_position["landmark100_fraction_of_CDS_UTR"] = np.where(
    top5_position["CDS_UTR_CLIP_total"] > 0,
    top5_position["landmark100_CLIP_count"] /
    top5_position["CDS_UTR_CLIP_total"],
    np.nan
)

top5_position["start100_fraction_of_CDS_UTR"] = np.where(
    top5_position["CDS_UTR_CLIP_total"] > 0,
    top5_position["start100_CLIP_count"] /
    top5_position["CDS_UTR_CLIP_total"],
    np.nan
)

top5_position["stop100_fraction_of_CDS_UTR"] = np.where(
    top5_position["CDS_UTR_CLIP_total"] > 0,
    top5_position["stop100_CLIP_count"] /
    top5_position["CDS_UTR_CLIP_total"],
    np.nan
)

startstop_corr_rows = []

for col in startstop_features:
    temp = top5_position[[col, "delta_TE"]].replace([np.inf, -np.inf], np.nan).dropna()

    startstop_corr_rows.append({
        "feature": col,
        "pearson_with_delta_TE": temp[col].corr(temp["delta_TE"], method="pearson"),
        "spearman_with_delta_TE": temp[col].corr(temp["delta_TE"], method="spearman"),
        "n_genes": len(temp)
    })

startstop_feature_correlations = pd.DataFrame(startstop_corr_rows)
startstop_feature_correlations
````

````python
# ============================================================
# Motif-context analysis for LIN28A CLIP reads
# ============================================================
# Goal:
# Among top 5% LIN28A-bound genes, compare high_delta_TE vs low_delta_TE
# in terms of LIN28A motif-containing CLIP fragments.
#
# Motifs:
# - AAGNNG-like: AAG[ACGT][ACGT]G
# - AAGNG-like : AAG[ACGT]G
# - UGUG-like  : TGTG in DNA sequence
#
# Note:
# This is a sequence-context analysis, not a full RNA structure prediction.
# A simple local complementarity score is included as a rough structure proxy.
# ============================================================

from pathlib import Path
import re
import gc
import pandas as pd
import numpy as np
from scipy.stats import mannwhitneyu
import matplotlib.pyplot as plt

# -----------------------------
# 0. Paths
# -----------------------------
DATA_DIR = Path("/home/ssy010622/bioinfo1_project/binfo1-work")

ASSIGN_DIR = DATA_DIR / "clip_read_assignments"
assign_file = ASSIGN_DIR / "CLIP-35L33G.bam.featureCounts"
clip_bam = DATA_DIR / "CLIP-35L33G.bam"

# Most recent table containing top5 + effect_group + region/position features
top5_file = DATA_DIR / "07_top5_binders_with_start_stop_CLIP_features.tsv"

OUT_DIR = DATA_DIR / "motif_context_results"
OUT_DIR.mkdir(exist_ok=True)

print("assign_file exists:", assign_file.exists())
print("clip_bam exists:", clip_bam.exists())
print("top5_file exists:", top5_file.exists())
print("OUT_DIR:", OUT_DIR)

# -----------------------------
# 1. Load top5 strong binder table
# -----------------------------
try:
    top5_position
    top5 = top5_position.copy()
    print("Using existing top5_position from notebook memory.")
except NameError:
    top5 = pd.read_csv(top5_file, sep="\t")
    print("Loaded:", top5_file)

# Ensure gene_id_no_version
if "gene_id_no_version" not in top5.columns:
    top5["gene_id_no_version"] = top5["Geneid"].astype(str).str.replace(r"\.\d+$", "", regex=True)

# Ensure effect_group
if "effect_group" not in top5.columns:
    q25 = top5["delta_TE"].quantile(0.25)
    q75 = top5["delta_TE"].quantile(0.75)
    top5["effect_group"] = "middle"
    top5.loc[top5["delta_TE"] <= q25, "effect_group"] = "low_delta_TE"
    top5.loc[top5["delta_TE"] >= q75, "effect_group"] = "high_delta_TE"

top5["effect_group"] = pd.Categorical(
    top5["effect_group"],
    categories=["low_delta_TE", "middle", "high_delta_TE"],
    ordered=True
)

print("\nTop5 effect group counts:")
display(top5["effect_group"].value_counts().reindex(["low_delta_TE", "middle", "high_delta_TE"]))

# Use all top5 genes for correlation, but compare high vs low later
target_geneids_version = set(top5["Geneid"].dropna().astype(str))
target_geneids_no_version = set(top5["gene_id_no_version"].dropna().astype(str))

# Map any ID format to canonical Geneid with version
id_to_canonical = {}

for gid in top5["Geneid"].dropna().astype(str):
    id_to_canonical[gid] = gid
    id_to_canonical[re.sub(r"\.\d+$", "", gid)] = gid

geneid_to_group = dict(zip(top5["Geneid"].astype(str), top5["effect_group"].astype(str)))
geneid_to_name = dict(zip(top5["Geneid"].astype(str), top5["gene_name"].astype(str)))

print("Number of target top5 genes:", len(target_geneids_version))

# -----------------------------
# 2. Parse featureCounts per-read assignment file
# -----------------------------
# assignment file format observed:
# read_id    status    n_features    gene_ids
#
# Example:
# SRR458758.xxxxx    Unassigned_NoFeatures    -1    NA
#
# Assigned rows should look like:
# SRR458758.xxxxx    Assigned    1    ENSMUSG...
# -----------------------------

def split_gene_ids(x):
    if pd.isna(x):
        return []
    x = str(x)
    if x == "NA":
        return []
    # featureCounts may separate multiple IDs by comma or semicolon
    parts = re.split(r"[,;]", x)
    return [p.strip() for p in parts if p.strip()]

read_to_genes = {}
gene_assignment_counts = {gid: 0 for gid in target_geneids_version}

chunk_size = 1_000_000
n_lines = 0
n_assigned_lines = 0
n_target_lines = 0

print("\nScanning assignment file in chunks...")

reader = pd.read_csv(
    assign_file,
    sep="\t",
    header=None,
    names=["read_id", "status", "n_features", "gene_ids"],
    dtype=str,
    chunksize=chunk_size
)

for chunk_i, chunk in enumerate(reader, start=1):
    n_lines += len(chunk)
    
    assigned = chunk[chunk["status"] == "Assigned"]
    n_assigned_lines += len(assigned)
    
    for row in assigned.itertuples(index=False):
        read_id = row.read_id
        gene_ids = split_gene_ids(row.gene_ids)
        
        matched = []
        for gid in gene_ids:
            gid_no_version = re.sub(r"\.\d+$", "", gid)
            
            if gid in id_to_canonical:
                matched.append(id_to_canonical[gid])
            elif gid_no_version in id_to_canonical:
                matched.append(id_to_canonical[gid_no_version])
        
        if len(matched) > 0:
            # remove duplicates while preserving order
            matched = list(dict.fromkeys(matched))
            read_to_genes[read_id] = matched
            n_target_lines += 1
            for mg in matched:
                gene_assignment_counts[mg] = gene_assignment_counts.get(mg, 0) + 1
    
    if chunk_i % 5 == 0:
        print(
            f"chunk {chunk_i} | lines scanned: {n_lines:,} | "
            f"assigned lines: {n_assigned_lines:,} | target reads: {len(read_to_genes):,}"
        )
        gc.collect()

print("\nFinished scanning assignment file.")
print("Total lines scanned:", f"{n_lines:,}")
print("Assigned lines:", f"{n_assigned_lines:,}")
print("Reads assigned to top5 genes:", f"{len(read_to_genes):,}")

# Save read-to-gene assignment summary, not the huge read_to_genes dict
gene_assignment_summary = pd.DataFrame({
    "Geneid": list(gene_assignment_counts.keys()),
    "assignment_file_read_count": list(gene_assignment_counts.values())
})

gene_assignment_summary["gene_name"] = gene_assignment_summary["Geneid"].map(geneid_to_name)
gene_assignment_summary["effect_group"] = gene_assignment_summary["Geneid"].map(geneid_to_group)

gene_assignment_summary.to_csv(
    OUT_DIR / "top5_gene_assignment_counts_from_featureCounts.tsv",
    sep="\t",
    index=False
)

display(gene_assignment_summary.sort_values("assignment_file_read_count", ascending=False).head(10))

# -----------------------------
# 3. Motif functions
# -----------------------------
motif_aagNNg = re.compile(r"AAG[ACGT][ACGT]G")
motif_aagNg  = re.compile(r"AAG[ACGT]G")
motif_ugug   = re.compile(r"TGTG")

comp_table = str.maketrans("ACGTNacgtn", "TGCANtgcan")

def revcomp(seq):
    return seq.translate(comp_table)[::-1].upper()

def motif_flags(seq):
    """
    Scan both read sequence and reverse complement.
    This makes motif detection orientation-agnostic.
    """
    if seq is None:
        seq = ""
    
    seq = seq.upper().replace("U", "T")
    rc = revcomp(seq)
    
    def has(regex, s):
        return bool(regex.search(s))
    
    aagNNg_fwd = has(motif_aagNNg, seq)
    aagNg_fwd  = has(motif_aagNg, seq)
    ugug_fwd   = has(motif_ugug, seq)
    
    aagNNg_either = aagNNg_fwd or has(motif_aagNNg, rc)
    aagNg_either  = aagNg_fwd  or has(motif_aagNg, rc)
    ugug_either   = ugug_fwd   or has(motif_ugug, rc)
    
    aag_family_either = aagNNg_either or aagNg_either
    any_motif_either = aag_family_either or ugug_either
    
    return {
        "has_AAGNNG": int(aagNNg_either),
        "has_AAGNG": int(aagNg_either),
        "has_UGUG": int(ugug_either),
        "has_AAG_family": int(aag_family_either),
        "has_any_LIN28A_motif": int(any_motif_either)
    }

def gc_fraction(seq):
    seq = seq.upper().replace("U", "T")
    if len(seq) == 0:
        return np.nan
    gc_count = seq.count("G") + seq.count("C")
    return gc_count / len(seq)

def purine_fraction(seq):
    seq = seq.upper().replace("U", "T")
    if len(seq) == 0:
        return np.nan
    purine_count = seq.count("A") + seq.count("G")
    return purine_count / len(seq)

def local_complementarity_proxy(seq, flank=5):
    """
    Very rough structure proxy:
    For each LIN28A-like motif occurrence, compare up to `flank` bases upstream
    with up to `flank` bases downstream in reverse-complement orientation.
    
    This is NOT RNAfold. It is only a lightweight sequence-based proxy for
    local hairpin-forming potential around motif-containing reads.
    """
    seq = seq.upper().replace("U", "T")
    if len(seq) < 8:
        return np.nan
    
    motif_regexes = [motif_aagNNg, motif_aagNg, motif_ugug]
    scores = []
    
    for regex in motif_regexes:
        for m in regex.finditer(seq):
            s, e = m.start(), m.end()
            left = seq[max(0, s - flank):s]
            right = seq[e:min(len(seq), e + flank)]
            
            k = min(len(left), len(right))
            if k == 0:
                continue
            
            left = left[-k:]
            right = right[:k]
            right_rc = revcomp(right)
            
            matches = sum(1 for a, b in zip(left, right_rc) if a == b)
            scores.append(matches / k)
    
    if len(scores) == 0:
        return 0.0
    
    return max(scores)

# -----------------------------
# 4. Scan BAM and collect motif stats per gene
# -----------------------------
try:
    import pysam
except ImportError as e:
    raise ImportError("pysam is required for this analysis.") from e

# initialize stats per gene
stats = {}
for gid in target_geneids_version:
    stats[gid] = {
        "Geneid": gid,
        "gene_name": geneid_to_name.get(gid, np.nan),
        "effect_group": geneid_to_group.get(gid, np.nan),
        "motif_total_reads": 0,
        "AAGNNG_reads": 0,
        "AAGNG_reads": 0,
        "UGUG_reads": 0,
        "AAG_family_reads": 0,
        "any_LIN28A_motif_reads": 0,
        "sum_read_length": 0,
        "sum_GC_fraction": 0.0,
        "sum_purine_fraction": 0.0,
        "sum_local_complementarity_proxy": 0.0
    }

seen_reads = set()
processed_target_reads = 0
skipped_no_seq = 0

print("\nScanning BAM for sequences of target reads...")

bam = pysam.AlignmentFile(str(clip_bam), "rb")

for i, read in enumerate(bam.fetch(until_eof=True), start=1):
    read_id = read.query_name
    
    if read_id not in read_to_genes:
        continue
    
    # Avoid counting the same read name repeatedly across multiple alignments
    if read_id in seen_reads:
        continue
    
    seen_reads.add(read_id)
    
    seq = read.query_sequence
    if seq is None:
        skipped_no_seq += 1
        continue
    
    seq = seq.upper().replace("U", "T")
    flags = motif_flags(seq)
    gc_frac = gc_fraction(seq)
    pur_frac = purine_fraction(seq)
    comp_proxy = local_complementarity_proxy(seq, flank=5)
    
    for gid in read_to_genes[read_id]:
        if gid not in stats:
            continue
        
        stats[gid]["motif_total_reads"] += 1
        stats[gid]["AAGNNG_reads"] += flags["has_AAGNNG"]
        stats[gid]["AAGNG_reads"] += flags["has_AAGNG"]
        stats[gid]["UGUG_reads"] += flags["has_UGUG"]
        stats[gid]["AAG_family_reads"] += flags["has_AAG_family"]
        stats[gid]["any_LIN28A_motif_reads"] += flags["has_any_LIN28A_motif"]
        stats[gid]["sum_read_length"] += len(seq)
        
        if not np.isnan(gc_frac):
            stats[gid]["sum_GC_fraction"] += gc_frac
        if not np.isnan(pur_frac):
            stats[gid]["sum_purine_fraction"] += pur_frac
        if not np.isnan(comp_proxy):
            stats[gid]["sum_local_complementarity_proxy"] += comp_proxy
    
    processed_target_reads += 1
    
    if processed_target_reads % 100000 == 0:
        print(f"processed target reads: {processed_target_reads:,}")

bam.close()

print("\nFinished BAM scan.")
print("Unique target reads processed:", f"{processed_target_reads:,}")
print("Skipped reads with no sequence:", skipped_no_seq)

# Free large read dictionary if memory is tight
# del read_to_genes
gc.collect()

# -----------------------------
# 5. Build gene-level motif feature table
# -----------------------------
motif_df = pd.DataFrame(list(stats.values()))

# Fractions
motif_df["AAGNNG_fraction"] = motif_df["AAGNNG_reads"] / motif_df["motif_total_reads"].replace(0, np.nan)
motif_df["AAGNG_fraction"] = motif_df["AAGNG_reads"] / motif_df["motif_total_reads"].replace(0, np.nan)
motif_df["UGUG_fraction"] = motif_df["UGUG_reads"] / motif_df["motif_total_reads"].replace(0, np.nan)
motif_df["AAG_family_fraction"] = motif_df["AAG_family_reads"] / motif_df["motif_total_reads"].replace(0, np.nan)
motif_df["any_LIN28A_motif_fraction"] = motif_df["any_LIN28A_motif_reads"] / motif_df["motif_total_reads"].replace(0, np.nan)

motif_df["mean_read_length"] = motif_df["sum_read_length"] / motif_df["motif_total_reads"].replace(0, np.nan)
motif_df["mean_GC_fraction"] = motif_df["sum_GC_fraction"] / motif_df["motif_total_reads"].replace(0, np.nan)
motif_df["mean_purine_fraction"] = motif_df["sum_purine_fraction"] / motif_df["motif_total_reads"].replace(0, np.nan)
motif_df["mean_local_complementarity_proxy"] = (
    motif_df["sum_local_complementarity_proxy"] / motif_df["motif_total_reads"].replace(0, np.nan)
)

# Merge with top5 metadata
motif_gene_table = top5.merge(
    motif_df.drop(columns=["gene_name", "effect_group"], errors="ignore"),
    on="Geneid",
    how="left"
)

# Fill missing motif counts with 0
count_cols = [
    "motif_total_reads",
    "AAGNNG_reads",
    "AAGNG_reads",
    "UGUG_reads",
    "AAG_family_reads",
    "any_LIN28A_motif_reads"
]
for c in count_cols:
    motif_gene_table[c] = motif_gene_table[c].fillna(0)

# Save full table
motif_gene_table_out = OUT_DIR / "top5_LIN28A_binders_motif_context_table.tsv"
motif_gene_table.to_csv(motif_gene_table_out, sep="\t", index=False)

print("\nSaved motif gene table:")
print(motif_gene_table_out)

display(
    motif_gene_table[
        [
            "gene_name", "effect_group", "delta_TE", "CLIP_enrichment",
            "motif_total_reads", "any_LIN28A_motif_fraction",
            "AAG_family_fraction", "UGUG_fraction",
            "mean_GC_fraction", "mean_purine_fraction",
            "mean_local_complementarity_proxy"
        ]
    ].sort_values("delta_TE", ascending=False).head(20)
)

# -----------------------------
# 6. Compare high_delta_TE vs low_delta_TE
# -----------------------------
# Require enough assigned reads per gene for stable motif fraction
MIN_READS = 20

motif_analysis = motif_gene_table[
    (motif_gene_table["effect_group"].isin(["low_delta_TE", "high_delta_TE"])) &
    (motif_gene_table["motif_total_reads"] >= MIN_READS)
].copy()

print("\nMotif analysis genes after MIN_READS filter:", len(motif_analysis))
print(motif_analysis["effect_group"].value_counts())

features_to_test = [
    "motif_total_reads",
    "AAGNNG_fraction",
    "AAGNG_fraction",
    "UGUG_fraction",
    "AAG_family_fraction",
    "any_LIN28A_motif_fraction",
    "mean_GC_fraction",
    "mean_purine_fraction",
    "mean_local_complementarity_proxy",
    "mean_read_length"
]

motif_test_rows = []

for col in features_to_test:
    low_values = motif_analysis.loc[
        motif_analysis["effect_group"] == "low_delta_TE", col
    ].replace([np.inf, -np.inf], np.nan).dropna()
    
    high_values = motif_analysis.loc[
        motif_analysis["effect_group"] == "high_delta_TE", col
    ].replace([np.inf, -np.inf], np.nan).dropna()
    
    if len(low_values) == 0 or len(high_values) == 0:
        pval = np.nan
    else:
        pval = mannwhitneyu(high_values, low_values, alternative="two-sided").pvalue
    
    motif_test_rows.append({
        "feature": col,
        "n_low": len(low_values),
        "n_high": len(high_values),
        "median_low": low_values.median() if len(low_values) > 0 else np.nan,
        "median_high": high_values.median() if len(high_values) > 0 else np.nan,
        "median_difference_high_minus_low": (
            high_values.median() - low_values.median()
            if len(low_values) > 0 and len(high_values) > 0 else np.nan
        ),
        "MWU_pvalue": pval
    })

motif_feature_tests = pd.DataFrame(motif_test_rows)

print("\nHigh vs low delta_TE motif feature tests:")
display(motif_feature_tests)

motif_feature_tests.to_csv(
    OUT_DIR / "motif_feature_tests_high_vs_low_deltaTE.tsv",
    sep="\t",
    index=False
)

# -----------------------------
# 7. Correlation within all top5 binders
# -----------------------------
motif_corr_rows = []

for col in features_to_test:
    temp = motif_gene_table[
        (motif_gene_table["motif_total_reads"] >= MIN_READS)
    ][[col, "delta_TE"]].replace([np.inf, -np.inf], np.nan).dropna()
    
    motif_corr_rows.append({
        "feature": col,
        "pearson_with_delta_TE": temp[col].corr(temp["delta_TE"], method="pearson"),
        "spearman_with_delta_TE": temp[col].corr(temp["delta_TE"], method="spearman"),
        "n_genes": len(temp)
    })

motif_feature_correlations = pd.DataFrame(motif_corr_rows)

print("\nMotif feature correlations with delta_TE among top5 binders:")
display(motif_feature_correlations)

motif_feature_correlations.to_csv(
    OUT_DIR / "motif_feature_correlations_with_deltaTE_top5.tsv",
    sep="\t",
    index=False
)

# -----------------------------
# 8. Plot key motif features
# -----------------------------
plot_features = [
    "any_LIN28A_motif_fraction",
    "AAG_family_fraction",
    "UGUG_fraction",
    "mean_local_complementarity_proxy"
]

for col in plot_features:
    plt.figure(figsize=(5, 4))
    
    motif_analysis.boxplot(
        column=col,
        by="effect_group",
        grid=False
    )
    
    plt.xlabel("")
    plt.ylabel(col)
    plt.title(f"{col}: high vs low delta_TE")
    plt.suptitle("")
    plt.xticks(rotation=20, ha="right")
    plt.tight_layout()
    
    out_png = OUT_DIR / f"motif_{col}_high_vs_low_deltaTE.png"
    plt.savefig(out_png, dpi=200)
    plt.show()

print("\nDone.")
print("Output directory:", OUT_DIR)
````

````python
# ============================================================
# Adjusted motif-context analysis
# ============================================================
# Question:
# Among strong LIN28A binders, are local motif features associated with delta_TE
# after adjusting for total CLIP signal, number of assigned CLIP reads,
# read length, RNA abundance, and baseline RPF abundance?
# ============================================================

from pathlib import Path
import pandas as pd
import numpy as np
from scipy.stats import pearsonr, spearmanr, mannwhitneyu
import matplotlib.pyplot as plt

DATA_DIR = Path("/home/ssy010622/bioinfo1_project/binfo1-work")
OUT_DIR = DATA_DIR / "motif_context_results"
OUT_DIR.mkdir(exist_ok=True)

motif_table_file = OUT_DIR / "top5_LIN28A_binders_motif_context_table.tsv"

try:
    motif_gene_table
    motif_df = motif_gene_table.copy()
    print("Using motif_gene_table from memory.")
except NameError:
    motif_df = pd.read_csv(motif_table_file, sep="\t")
    print("Loaded:", motif_table_file)

# Minimum number of assigned CLIP reads per gene
MIN_READS = 20

motif_df = motif_df[motif_df["motif_total_reads"] >= MIN_READS].copy()

# Basic transformed covariates
motif_df["log2_motif_total_reads"] = np.log2(motif_df["motif_total_reads"] + 1)

if "RNA-siLuc_CPM" in motif_df.columns:
    motif_df["log2_RNA_siLuc_CPM"] = np.log2(motif_df["RNA-siLuc_CPM"] + 0.5)

if "RPF-siLuc_CPM" in motif_df.columns:
    motif_df["log2_RPF_siLuc_CPM"] = np.log2(motif_df["RPF-siLuc_CPM"] + 0.5)

# Optional basal TE if not already present
if "basal_TE_siLuc" not in motif_df.columns:
    if ("RPF-siLuc_CPM" in motif_df.columns) and ("RNA-siLuc_CPM" in motif_df.columns):
        motif_df["basal_TE_siLuc"] = np.log2(
            (motif_df["RPF-siLuc_CPM"] + 0.5) /
            (motif_df["RNA-siLuc_CPM"] + 0.5)
        )

# Candidate motif/context features
motif_features = [
    "AAGNNG_fraction",
    "AAGNG_fraction",
    "UGUG_fraction",
    "AAG_family_fraction",
    "any_LIN28A_motif_fraction",
    "mean_GC_fraction",
    "mean_purine_fraction",
    "mean_local_complementarity_proxy"
]

# Covariates to adjust for
candidate_covariates = [
    "CLIP_enrichment",
    "log2_motif_total_reads",
    "mean_read_length",
    "log2_RNA_siLuc_CPM",
    "log2_RPF_siLuc_CPM",
    "basal_TE_siLuc"
]

covariates = [c for c in candidate_covariates if c in motif_df.columns]

print("Number of genes used:", len(motif_df))
print("Covariates:")
print(covariates)

# -----------------------------
# Helper functions
# -----------------------------
def residualize_vector(y, X):
    """
    Return residuals of y after linear regression on X.
    X should already include intercept if desired.
    """
    y = np.asarray(y, dtype=float)
    X = np.asarray(X, dtype=float)

    valid = np.isfinite(y) & np.all(np.isfinite(X), axis=1)

    residuals = np.full(len(y), np.nan)

    if valid.sum() < X.shape[1] + 2:
        return residuals

    beta = np.linalg.lstsq(X[valid], y[valid], rcond=None)[0]
    residuals[valid] = y[valid] - X[valid].dot(beta)

    return residuals

def bh_fdr(pvalues):
    """
    Benjamini-Hochberg FDR correction.
    """
    pvalues = np.asarray(pvalues, dtype=float)
    qvalues = np.full_like(pvalues, np.nan)

    valid = np.isfinite(pvalues)
    p = pvalues[valid]
    n = len(p)

    if n == 0:
        return qvalues

    order = np.argsort(p)
    ranked_p = p[order]

    q = ranked_p * n / (np.arange(1, n + 1))
    q = np.minimum.accumulate(q[::-1])[::-1]
    q = np.clip(q, 0, 1)

    q_original_order = np.empty(n)
    q_original_order[order] = q

    qvalues[valid] = q_original_order
    return qvalues

# -----------------------------
# 1. Adjusted residual correlation with delta_TE
# -----------------------------
adjusted_corr_rows = []

for feature in motif_features:
    needed_cols = [feature, "delta_TE"] + covariates
    sub = motif_df[needed_cols].replace([np.inf, -np.inf], np.nan).dropna().copy()

    if len(sub) < 30:
        adjusted_corr_rows.append({
            "feature": feature,
            "n_genes": len(sub),
            "partial_pearson_r": np.nan,
            "partial_pearson_p": np.nan,
            "partial_spearman_r": np.nan,
            "partial_spearman_p": np.nan
        })
        continue

    X = sub[covariates].to_numpy(dtype=float)
    X = np.column_stack([np.ones(len(sub)), X])

    y_resid = residualize_vector(sub["delta_TE"].to_numpy(dtype=float), X)
    x_resid = residualize_vector(sub[feature].to_numpy(dtype=float), X)

    valid = np.isfinite(y_resid) & np.isfinite(x_resid)

    pearson_result = pearsonr(x_resid[valid], y_resid[valid])
    spearman_result = spearmanr(x_resid[valid], y_resid[valid])

    adjusted_corr_rows.append({
        "feature": feature,
        "n_genes": int(valid.sum()),
        "partial_pearson_r": pearson_result.statistic,
        "partial_pearson_p": pearson_result.pvalue,
        "partial_spearman_r": spearman_result.statistic,
        "partial_spearman_p": spearman_result.pvalue
    })

adjusted_motif_correlations = pd.DataFrame(adjusted_corr_rows)
adjusted_motif_correlations["partial_pearson_FDR"] = bh_fdr(
    adjusted_motif_correlations["partial_pearson_p"]
)
adjusted_motif_correlations["partial_spearman_FDR"] = bh_fdr(
    adjusted_motif_correlations["partial_spearman_p"]
)

display(adjusted_motif_correlations)

adjusted_motif_correlations.to_csv(
    OUT_DIR / "adjusted_motif_correlations_with_deltaTE.tsv",
    sep="\t",
    index=False
)

# -----------------------------
# 2. Adjusted high vs low group comparison
# -----------------------------
extreme_df = motif_df[
    motif_df["effect_group"].isin(["low_delta_TE", "high_delta_TE"])
].copy()

adjusted_group_rows = []

for feature in motif_features:
    needed_cols = [feature, "effect_group"] + covariates
    sub = extreme_df[needed_cols].replace([np.inf, -np.inf], np.nan).dropna().copy()

    if len(sub) < 30:
        adjusted_group_rows.append({
            "feature": feature,
            "n_low": np.nan,
            "n_high": np.nan,
            "median_residual_low": np.nan,
            "median_residual_high": np.nan,
            "median_residual_difference_high_minus_low": np.nan,
            "MWU_pvalue": np.nan
        })
        continue

    X = sub[covariates].to_numpy(dtype=float)
    X = np.column_stack([np.ones(len(sub)), X])

    feature_resid = residualize_vector(sub[feature].to_numpy(dtype=float), X)
    sub[f"{feature}_adjusted_residual"] = feature_resid

    low_values = sub.loc[
        sub["effect_group"] == "low_delta_TE",
        f"{feature}_adjusted_residual"
    ].dropna()

    high_values = sub.loc[
        sub["effect_group"] == "high_delta_TE",
        f"{feature}_adjusted_residual"
    ].dropna()

    if len(low_values) == 0 or len(high_values) == 0:
        pval = np.nan
    else:
        pval = mannwhitneyu(high_values, low_values, alternative="two-sided").pvalue

    adjusted_group_rows.append({
        "feature": feature,
        "n_low": len(low_values),
        "n_high": len(high_values),
        "median_residual_low": low_values.median(),
        "median_residual_high": high_values.median(),
        "median_residual_difference_high_minus_low": high_values.median() - low_values.median(),
        "MWU_pvalue": pval
    })

adjusted_group_tests = pd.DataFrame(adjusted_group_rows)
adjusted_group_tests["MWU_FDR"] = bh_fdr(adjusted_group_tests["MWU_pvalue"])

display(adjusted_group_tests)

adjusted_group_tests.to_csv(
    OUT_DIR / "adjusted_motif_group_tests_high_vs_low.tsv",
    sep="\t",
    index=False
)

# -----------------------------
# 3. Make adjusted residual plots for key features
# -----------------------------
key_plot_features = [
    "AAG_family_fraction",
    "UGUG_fraction",
    "any_LIN28A_motif_fraction",
    "mean_local_complementarity_proxy"
]

for feature in key_plot_features:
    needed_cols = [feature, "effect_group"] + covariates
    sub = extreme_df[needed_cols].replace([np.inf, -np.inf], np.nan).dropna().copy()

    X = sub[covariates].to_numpy(dtype=float)
    X = np.column_stack([np.ones(len(sub)), X])

    sub[f"{feature}_adjusted_residual"] = residualize_vector(
        sub[feature].to_numpy(dtype=float),
        X
    )

    sub = sub[sub["effect_group"].isin(["low_delta_TE", "high_delta_TE"])].copy()

    plt.figure(figsize=(5, 4))

    sub.boxplot(
        column=f"{feature}_adjusted_residual",
        by="effect_group",
        grid=False
    )

    plt.axhline(0, linestyle="--")
    plt.xlabel("")
    plt.ylabel(f"Adjusted residual of {feature}")
    plt.title(f"Adjusted {feature}: high vs low delta_TE")
    plt.suptitle("")
    plt.xticks(rotation=20, ha="right")
    plt.tight_layout()

    out_png = OUT_DIR / f"adjusted_{feature}_high_vs_low_deltaTE.png"
    plt.savefig(out_png, dpi=200)
    plt.show()

print("Saved outputs to:", OUT_DIR)
````

### Within strongly LIN28A-bound mRNAs, functional translational derepression was not explained by broad CDS/UTR binding bias or start/stop-proximal binding. Unexpectedly, high-delta_TE targets showed lower AAG-family motif fraction and slightly higher UGUG motif fraction among assigned CLIP fragments, suggesting that functional LIN28A repression may depend on a distinct local motif context rather than canonical motif abundance alone.

````python
# ============================================================
# Report-ready figures and tables for LIN28A project
# ============================================================

from pathlib import Path
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import mannwhitneyu

# -----------------------------
# Paths
# -----------------------------
DATA_DIR = Path("/home/ssy010622/bioinfo1_project/binfo1-work")

FIG_DIR = DATA_DIR / "report_figures"
TABLE_DIR = DATA_DIR / "report_tables"

FIG_DIR.mkdir(exist_ok=True)
TABLE_DIR.mkdir(exist_ok=True)

binned_file = DATA_DIR / "02_paperlike_no_histone_binned.tsv"
top5_position_file = DATA_DIR / "07_top5_binders_with_start_stop_CLIP_features.tsv"

motif_dir = DATA_DIR / "motif_context_results"
motif_table_file = motif_dir / "top5_LIN28A_binders_motif_context_table.tsv"
adjusted_corr_file = motif_dir / "adjusted_motif_correlations_with_deltaTE.tsv"
adjusted_group_file = motif_dir / "adjusted_motif_group_tests_high_vs_low.tsv"

print("Loading files...")
print("binned_file:", binned_file.exists(), binned_file)
print("top5_position_file:", top5_position_file.exists(), top5_position_file)
print("motif_table_file:", motif_table_file.exists(), motif_table_file)
print("adjusted_corr_file:", adjusted_corr_file.exists(), adjusted_corr_file)
print("adjusted_group_file:", adjusted_group_file.exists(), adjusted_group_file)

df = pd.read_csv(binned_file, sep="\t")
top5 = pd.read_csv(top5_position_file, sep="\t")
motif = pd.read_csv(motif_table_file, sep="\t")
adj_corr = pd.read_csv(adjusted_corr_file, sep="\t")
adj_group = pd.read_csv(adjusted_group_file, sep="\t")

# -----------------------------
# Style
# -----------------------------
plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "font.size": 9,
    "axes.labelsize": 10,
    "axes.titlesize": 11,
    "xtick.labelsize": 8.5,
    "ytick.labelsize": 8.5,
    "legend.fontsize": 8.5,
    "figure.dpi": 120,
    "savefig.dpi": 300,
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
    "axes.spines.top": False,
    "axes.spines.right": False,
})

COLORS = {
    "background": "#BDBDBD",
    "bottom 50%": "#4C78A8",
    "20-50%": "#F58518",
    "5-20%": "#54A24B",
    "top 5%": "#E45756",
    "low_delta_TE": "#4C78A8",
    "middle": "#BDBDBD",
    "high_delta_TE": "#E45756",
    "positive": "#0072B2",
    "negative": "#D55E00",
    "neutral": "#999999"
}

def save_figure(fig, name):
    fig.savefig(FIG_DIR / f"{name}.png", bbox_inches="tight")
    fig.savefig(FIG_DIR / f"{name}.pdf", bbox_inches="tight")
    fig.savefig(FIG_DIR / f"{name}.svg", bbox_inches="tight")
    print("Saved:", FIG_DIR / f"{name}.png")

def bh_fdr(pvalues):
    pvalues = np.asarray(pvalues, dtype=float)
    qvalues = np.full_like(pvalues, np.nan)
    valid = np.isfinite(pvalues)
    p = pvalues[valid]
    n = len(p)
    if n == 0:
        return qvalues
    order = np.argsort(p)
    ranked = p[order]
    q = ranked * n / np.arange(1, n + 1)
    q = np.minimum.accumulate(q[::-1])[::-1]
    q = np.clip(q, 0, 1)
    out = np.empty(n)
    out[order] = q
    qvalues[valid] = out
    return qvalues

def ensure_effect_group(table):
    table = table.copy()
    if "effect_group" not in table.columns:
        q25 = table["delta_TE"].quantile(0.25)
        q75 = table["delta_TE"].quantile(0.75)
        table["effect_group"] = "middle"
        table.loc[table["delta_TE"] <= q25, "effect_group"] = "low_delta_TE"
        table.loc[table["delta_TE"] >= q75, "effect_group"] = "high_delta_TE"
    table["effect_group"] = pd.Categorical(
        table["effect_group"],
        categories=["low_delta_TE", "middle", "high_delta_TE"],
        ordered=True
    )
    return table

top5 = ensure_effect_group(top5)
motif = ensure_effect_group(motif)

# Make sure group order is stable
binding_order = ["bottom 50%", "20-50%", "5-20%", "top 5%"]
df["LIN28A_binding_group"] = pd.Categorical(
    df["LIN28A_binding_group"],
    categories=binding_order,
    ordered=True
)

# ============================================================
# Table 1. Binding group summary
# ============================================================
table1_binding_summary = df.groupby("LIN28A_binding_group").agg(
    n_genes=("Geneid", "count"),
    median_CLIP_enrichment=("CLIP_enrichment", "median"),
    median_log2FC_RNA=("log2FC_RNA", "median"),
    median_log2FC_RPF=("log2FC_RPF", "median"),
    median_delta_TE=("delta_TE", "median"),
    mean_delta_TE=("delta_TE", "mean")
).reset_index()

table1_binding_summary.to_csv(
    TABLE_DIR / "Table1_binding_group_summary.tsv",
    sep="\t",
    index=False
)

display(table1_binding_summary)

# ============================================================
# Figure 1. CDF of delta_TE by LIN28A binding strength
# ============================================================
fig, ax = plt.subplots(figsize=(5.2, 4.1))

for group in binding_order:
    values = df.loc[df["LIN28A_binding_group"] == group, "delta_TE"].dropna().sort_values()
    y = np.arange(1, len(values) + 1) / len(values)
    median_val = values.median()
    label = f"{group} (n={len(values)}, median={median_val:.2f})"
    ax.step(values, y, where="post", lw=2.0, color=COLORS[group], label=label)

ax.axvline(0, color="black", lw=0.8, ls="--", alpha=0.6)
ax.set_xlabel(r"$\Delta$TE = log$_2$FC(RPF) - log$_2$FC(RNA)")
ax.set_ylabel("Cumulative fraction")
ax.set_title("LIN28A binding strength predicts translational derepression")
ax.legend(frameon=False, loc="lower right")
ax.set_xlim(df["delta_TE"].quantile(0.005), df["delta_TE"].quantile(0.995))
ax.set_ylim(0, 1.01)

save_figure(fig, "Figure1_deltaTE_CDF_by_LIN28A_binding")
plt.show()

# ============================================================
# Figure 2. Scatter showing high/low effect split among strong binders
# ============================================================
plot_df = df.copy()

high_ids = set(top5.loc[top5["effect_group"] == "high_delta_TE", "Geneid"].astype(str))
low_ids = set(top5.loc[top5["effect_group"] == "low_delta_TE", "Geneid"].astype(str))
middle_ids = set(top5.loc[top5["effect_group"] == "middle", "Geneid"].astype(str))

plot_df["plot_group"] = "background"
plot_df.loc[plot_df["Geneid"].astype(str).isin(middle_ids), "plot_group"] = "top5_middle"
plot_df.loc[plot_df["Geneid"].astype(str).isin(low_ids), "plot_group"] = "low_delta_TE"
plot_df.loc[plot_df["Geneid"].astype(str).isin(high_ids), "plot_group"] = "high_delta_TE"

fig, ax = plt.subplots(figsize=(5.4, 4.3))

# background
bg = plot_df[plot_df["plot_group"] == "background"]
ax.scatter(
    bg["CLIP_enrichment"],
    bg["delta_TE"],
    s=8,
    color="#D0D0D0",
    alpha=0.35,
    linewidths=0,
    label="Other expressed genes"
)

# middle top5
mid = plot_df[plot_df["plot_group"] == "top5_middle"]
ax.scatter(
    mid["CLIP_enrichment"],
    mid["delta_TE"],
    s=18,
    color="#A9A9A9",
    alpha=0.65,
    linewidths=0,
    label="Top 5% binders, middle"
)

# low and high
for group, label in [
    ("low_delta_TE", "Top 5% binders, low ΔTE"),
    ("high_delta_TE", "Top 5% binders, high ΔTE")
]:
    sub = plot_df[plot_df["plot_group"] == group]
    ax.scatter(
        sub["CLIP_enrichment"],
        sub["delta_TE"],
        s=24,
        color=COLORS[group],
        alpha=0.85,
        edgecolor="white",
        linewidth=0.25,
        label=label
    )

ax.axhline(0, color="black", lw=0.8, ls="--", alpha=0.6)
ax.set_xlabel("LIN28A CLIP enrichment")
ax.set_ylabel(r"$\Delta$TE")
ax.set_title("Strong LIN28A binders separate into high- and low-effect targets")
ax.legend(frameon=False, loc="best")

# label representative genes
label_table = top5.copy()
label_genes = (
    label_table[label_table["effect_group"] == "high_delta_TE"]
    .sort_values("delta_TE", ascending=False)
    .head(5)["gene_name"]
    .dropna()
    .astype(str)
    .tolist()
    +
    label_table[label_table["effect_group"] == "low_delta_TE"]
    .sort_values("delta_TE", ascending=True)
    .head(4)["gene_name"]
    .dropna()
    .astype(str)
    .tolist()
)

for _, row in plot_df[plot_df["gene_name"].astype(str).isin(label_genes)].iterrows():
    ax.text(
        row["CLIP_enrichment"] + 0.03,
        row["delta_TE"] + 0.03,
        str(row["gene_name"]),
        fontsize=7,
        alpha=0.9
    )

save_figure(fig, "Figure2_strong_binder_high_low_scatter")
plt.show()

# ============================================================
# Helper: Mann-Whitney rank-biserial effect size
# ============================================================
def mwu_rank_biserial(high_values, low_values):
    high_values = pd.Series(high_values).replace([np.inf, -np.inf], np.nan).dropna()
    low_values = pd.Series(low_values).replace([np.inf, -np.inf], np.nan).dropna()
    if len(high_values) == 0 or len(low_values) == 0:
        return np.nan, np.nan
    res = mannwhitneyu(high_values, low_values, alternative="two-sided")
    rank_biserial = 2 * res.statistic / (len(high_values) * len(low_values)) - 1
    return rank_biserial, res.pvalue

# ============================================================
# Table 2 + Figure 3. Coarse positional features
# ============================================================
# Create additional landmark features if missing
if "landmark100_CLIP_count" not in top5.columns:
    top5["landmark100_CLIP_count"] = top5["start100_CLIP_count"] + top5["stop100_CLIP_count"]

if "landmark100_fraction_of_CDS_UTR" not in top5.columns:
    top5["landmark100_fraction_of_CDS_UTR"] = np.where(
        top5["CDS_UTR_CLIP_total"] > 0,
        top5["landmark100_CLIP_count"] / top5["CDS_UTR_CLIP_total"],
        np.nan
    )

if "start100_fraction_of_CDS_UTR" not in top5.columns:
    top5["start100_fraction_of_CDS_UTR"] = np.where(
        top5["CDS_UTR_CLIP_total"] > 0,
        top5["start100_CLIP_count"] / top5["CDS_UTR_CLIP_total"],
        np.nan
    )

if "stop100_fraction_of_CDS_UTR" not in top5.columns:
    top5["stop100_fraction_of_CDS_UTR"] = np.where(
        top5["CDS_UTR_CLIP_total"] > 0,
        top5["stop100_CLIP_count"] / top5["CDS_UTR_CLIP_total"],
        np.nan
    )

coarse_features = [
    "CDS_CLIP_fraction",
    "CDS_vs_UTR_density_bias",
    "start100_CLIP_fraction",
    "start_vs_stop_density_bias",
    "landmark100_fraction_of_CDS_UTR",
    "start100_fraction_of_CDS_UTR",
    "stop100_fraction_of_CDS_UTR"
]

coarse_labels = {
    "CDS_CLIP_fraction": "CDS fraction",
    "CDS_vs_UTR_density_bias": "CDS/UTR density bias",
    "start100_CLIP_fraction": "Start/(start+stop)",
    "start_vs_stop_density_bias": "Start/stop density bias",
    "landmark100_fraction_of_CDS_UTR": "Start+stop windows / CDS+UTR",
    "start100_fraction_of_CDS_UTR": "Start window / CDS+UTR",
    "stop100_fraction_of_CDS_UTR": "Stop window / CDS+UTR"
}

extreme_top5 = top5[top5["effect_group"].isin(["low_delta_TE", "high_delta_TE"])].copy()

coarse_rows = []
for feature in coarse_features:
    if feature not in extreme_top5.columns:
        continue
    low = extreme_top5.loc[extreme_top5["effect_group"] == "low_delta_TE", feature]
    high = extreme_top5.loc[extreme_top5["effect_group"] == "high_delta_TE", feature]
    rb, p = mwu_rank_biserial(high, low)
    low_clean = pd.Series(low).replace([np.inf, -np.inf], np.nan).dropna()
    high_clean = pd.Series(high).replace([np.inf, -np.inf], np.nan).dropna()
    coarse_rows.append({
        "feature": feature,
        "label": coarse_labels.get(feature, feature),
        "n_low": len(low_clean),
        "n_high": len(high_clean),
        "median_low": low_clean.median(),
        "median_high": high_clean.median(),
        "median_difference_high_minus_low": high_clean.median() - low_clean.median(),
        "rank_biserial_effect_high_vs_low": rb,
        "MWU_pvalue": p
    })

table2_position_tests = pd.DataFrame(coarse_rows)
table2_position_tests["MWU_FDR"] = bh_fdr(table2_position_tests["MWU_pvalue"])

table2_position_tests.to_csv(
    TABLE_DIR / "Table2_coarse_position_feature_tests.tsv",
    sep="\t",
    index=False
)

display(table2_position_tests)

# Figure 3: lollipop of rank-biserial effects
plot_table = table2_position_tests.copy()
plot_table = plot_table.sort_values("rank_biserial_effect_high_vs_low")

fig, ax = plt.subplots(figsize=(5.3, 3.7))
y = np.arange(len(plot_table))
effects = plot_table["rank_biserial_effect_high_vs_low"].values

for yi, eff, fdr in zip(y, effects, plot_table["MWU_FDR"]):
    color = COLORS["neutral"]
    if np.isfinite(fdr) and fdr < 0.05:
        color = COLORS["positive"] if eff > 0 else COLORS["negative"]
    ax.hlines(yi, 0, eff, color=color, lw=2)
    ax.scatter(eff, yi, color=color, s=34, zorder=3)

ax.axvline(0, color="black", lw=0.8, ls="--")
ax.set_yticks(y)
ax.set_yticklabels(plot_table["label"])
ax.set_xlabel("Rank-biserial effect size\n(high ΔTE vs low ΔTE)")
ax.set_title("Broad positional features weakly explain high-effect targets")

for yi, eff, fdr in zip(y, effects, plot_table["MWU_FDR"]):
    ax.text(
        0.52 if eff >= 0 else -0.52,
        yi,
        f"FDR={fdr:.2g}",
        va="center",
        ha="right" if eff < 0 else "left",
        fontsize=7.5,
        color="#555555"
    )

ax.set_xlim(-0.65, 0.65)
save_figure(fig, "Figure3_coarse_position_feature_effects")
plt.show()

# ============================================================
# Table 3 raw motif tests + Figure 4 raw motif boxplots
# ============================================================
MIN_READS = 20
motif_analysis = motif[
    (motif["effect_group"].isin(["low_delta_TE", "high_delta_TE"])) &
    (motif["motif_total_reads"] >= MIN_READS)
].copy()

motif_features_raw = [
    "motif_total_reads",
    "AAGNNG_fraction",
    "AAGNG_fraction",
    "UGUG_fraction",
    "AAG_family_fraction",
    "any_LIN28A_motif_fraction",
    "mean_GC_fraction",
    "mean_purine_fraction",
    "mean_local_complementarity_proxy",
    "mean_read_length"
]

motif_rows = []
for feature in motif_features_raw:
    low = motif_analysis.loc[motif_analysis["effect_group"] == "low_delta_TE", feature]
    high = motif_analysis.loc[motif_analysis["effect_group"] == "high_delta_TE", feature]
    rb, p = mwu_rank_biserial(high, low)
    low_clean = pd.Series(low).replace([np.inf, -np.inf], np.nan).dropna()
    high_clean = pd.Series(high).replace([np.inf, -np.inf], np.nan).dropna()
    motif_rows.append({
        "feature": feature,
        "n_low": len(low_clean),
        "n_high": len(high_clean),
        "median_low": low_clean.median(),
        "median_high": high_clean.median(),
        "median_difference_high_minus_low": high_clean.median() - low_clean.median(),
        "rank_biserial_effect_high_vs_low": rb,
        "MWU_pvalue": p
    })

table3_raw_motif_tests = pd.DataFrame(motif_rows)
table3_raw_motif_tests["MWU_FDR"] = bh_fdr(table3_raw_motif_tests["MWU_pvalue"])

table3_raw_motif_tests.to_csv(
    TABLE_DIR / "Table3_raw_motif_feature_tests.tsv",
    sep="\t",
    index=False
)

display(table3_raw_motif_tests)

# Plot function
def box_jitter(ax, data, feature, ylabel, title, stats_table=None):
    groups = ["low_delta_TE", "high_delta_TE"]
    positions = [1, 2]
    arrays = [
        data.loc[data["effect_group"] == g, feature].replace([np.inf, -np.inf], np.nan).dropna().values
        for g in groups
    ]
    
    bp = ax.boxplot(
        arrays,
        positions=positions,
        widths=0.55,
        patch_artist=True,
        showfliers=False,
        medianprops={"color": "black", "linewidth": 1.2},
        boxprops={"linewidth": 1.0},
        whiskerprops={"linewidth": 1.0},
        capprops={"linewidth": 1.0}
    )
    
    for patch, g in zip(bp["boxes"], groups):
        patch.set_facecolor(COLORS[g])
        patch.set_alpha(0.45)
    
    rng = np.random.default_rng(10)
    for pos, values, g in zip(positions, arrays, groups):
        jitter = rng.normal(0, 0.045, size=len(values))
        ax.scatter(
            np.full(len(values), pos) + jitter,
            values,
            s=10,
            color=COLORS[g],
            alpha=0.55,
            linewidths=0
        )
    
    ax.set_xticks(positions)
    ax.set_xticklabels(["Low ΔTE", "High ΔTE"], rotation=0)
    ax.set_ylabel(ylabel)
    ax.set_title(title)
    
    if stats_table is not None:
        row = stats_table[stats_table["feature"] == feature]
        if len(row) == 1:
            p = row["MWU_pvalue"].iloc[0]
            fdr = row["MWU_FDR"].iloc[0]
            ax.text(
                0.02, 0.97,
                f"MWU p={p:.2g}\nFDR={fdr:.2g}",
                transform=ax.transAxes,
                ha="left",
                va="top",
                fontsize=8
            )

fig, axes = plt.subplots(1, 2, figsize=(7.0, 3.4))

box_jitter(
    axes[0],
    motif_analysis,
    "AAG_family_fraction",
    "Fraction of assigned CLIP reads",
    "AAG-family motif fraction",
    table3_raw_motif_tests
)

box_jitter(
    axes[1],
    motif_analysis,
    "UGUG_fraction",
    "Fraction of assigned CLIP reads",
    "UGUG-like motif fraction",
    table3_raw_motif_tests
)

axes[0].text(-0.18, 1.05, "A", transform=axes[0].transAxes, fontsize=13, fontweight="bold")
axes[1].text(-0.18, 1.05, "B", transform=axes[1].transAxes, fontsize=13, fontweight="bold")

fig.suptitle("Local LIN28A motif composition differs among strong binders", y=1.04, fontsize=12)
fig.tight_layout()

save_figure(fig, "Figure4_local_motif_composition_boxplots")
plt.show()

# ============================================================
# Figure 5. Adjusted motif partial correlations
# ============================================================
feature_label_map = {
    "AAGNNG_fraction": "AAGNNG",
    "AAGNG_fraction": "AAGNG",
    "UGUG_fraction": "UGUG",
    "AAG_family_fraction": "AAG family",
    "any_LIN28A_motif_fraction": "Any LIN28A motif",
    "mean_GC_fraction": "GC fraction",
    "mean_purine_fraction": "Purine fraction",
    "mean_local_complementarity_proxy": "Local complementarity proxy"
}

plot_adj = adj_corr.copy()
plot_adj["label"] = plot_adj["feature"].map(feature_label_map).fillna(plot_adj["feature"])
plot_adj = plot_adj.sort_values("partial_pearson_r")

fig, ax = plt.subplots(figsize=(5.2, 3.9))

y = np.arange(len(plot_adj))
x = plot_adj["partial_pearson_r"].values
fdr = plot_adj["partial_pearson_FDR"].values

for yi, xi, qi in zip(y, x, fdr):
    if np.isfinite(qi) and qi < 0.05:
        color = COLORS["positive"] if xi > 0 else COLORS["negative"]
    else:
        color = COLORS["neutral"]
    ax.hlines(yi, 0, xi, color=color, lw=2)
    ax.scatter(xi, yi, color=color, s=36, zorder=3)

ax.axvline(0, color="black", lw=0.8, ls="--")
ax.set_yticks(y)
ax.set_yticklabels(plot_adj["label"])
ax.set_xlabel("Adjusted partial Pearson r with ΔTE")
ax.set_title("Adjusted motif-context associations among strong LIN28A binders")

for yi, xi, qi in zip(y, x, fdr):
    ax.text(
        0.22 if xi >= 0 else -0.22,
        yi,
        f"FDR={qi:.2g}",
        va="center",
        ha="left" if xi >= 0 else "right",
        fontsize=7.5,
        color="#555555"
    )

ax.set_xlim(-0.30, 0.30)
save_figure(fig, "Figure5_adjusted_motif_partial_correlations")
plt.show()

# ============================================================
# Table 4. Adjusted motif summary
# ============================================================
table4_adjusted_motif_summary = adj_corr.merge(
    adj_group[
        [
            "feature",
            "n_low",
            "n_high",
            "median_residual_low",
            "median_residual_high",
            "median_residual_difference_high_minus_low",
            "MWU_pvalue",
            "MWU_FDR"
        ]
    ],
    on="feature",
    how="left"
)

table4_adjusted_motif_summary.to_csv(
    TABLE_DIR / "Table4_adjusted_motif_context_summary.tsv",
    sep="\t",
    index=False
)

display(table4_adjusted_motif_summary)

# ============================================================
# Supplemental example table: top high/low effect genes
# ============================================================
motif_cols = [
    "Geneid",
    "AAG_family_fraction",
    "UGUG_fraction",
    "any_LIN28A_motif_fraction",
    "motif_total_reads"
]

motif_small = motif[[c for c in motif_cols if c in motif.columns]].copy()

example_base = top5.merge(
    motif_small,
    on="Geneid",
    how="left",
    suffixes=("", "_motif")
)

example_cols = [
    "effect_group",
    "Geneid",
    "gene_name",
    "CLIP_enrichment",
    "log2FC_RNA",
    "log2FC_RPF",
    "delta_TE",
    "RNA-siLuc_CPM",
    "RPF-siLuc_CPM",
    "AAG_family_fraction",
    "UGUG_fraction",
    "any_LIN28A_motif_fraction",
    "motif_total_reads"
]

example_high = (
    example_base[example_base["effect_group"] == "high_delta_TE"]
    .sort_values("delta_TE", ascending=False)
    .head(12)
)

example_low = (
    example_base[example_base["effect_group"] == "low_delta_TE"]
    .sort_values("delta_TE", ascending=True)
    .head(12)
)

tableS_examples = pd.concat([example_high, example_low], axis=0)
tableS_examples = tableS_examples[[c for c in example_cols if c in tableS_examples.columns]]

tableS_examples.to_csv(
    TABLE_DIR / "TableS1_representative_high_low_effect_strong_binders.tsv",
    sep="\t",
    index=False
)

display(tableS_examples)

print("\nAll report figures saved to:", FIG_DIR)
print("All report tables saved to:", TABLE_DIR)
````
